# Create new data set by encoding catagorical value for analysis

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
data=pd.DataFrame()
data['price']=df['price']
data['area']=df['area']
data['bedroomes']=df['bedrooms']
data['bathrooms']=df['bathrooms']
data['stories']=df['stories']
data['price']=df['price']
label_encoder = LabelEncoder()
data['mainroad_encoded'] = label_encoder.fit_transform(df['mainroad'])
data['guestroom_encoded'] = label_encoder.fit_transform(df['guestroom'])
data['basement_encoded'] = label_encoder.fit_transform(df['basement'])
data['hotwaterheating_encoded'] = label_encoder.fit_transform(df['hotwaterheating'])
data['airconditioning_encoded'] = label_encoder.fit_transform(df['airconditioning'])
data['parking']=df['parking']
data['prefarea_encoded'] = label_encoder.fit_transform(df['prefarea'])
data['furnishingstatus_encoded'] = label_encoder.fit_transform(df['furnishingstatus'])
```

```python
display(data)
```

# Split into Train and Test

```python
X=data.drop('price',axis=1)
y=data['price']
# Split: 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,    # 20% for test
    random_state=20   # Same result every time
)

print(f"\nWe will train with: {len(X_train)} houses")
print(f"We will test with: {len(X_test)} houses")
```

```python
display(X_train)
```

```python
display(X_test)
```

# Apply StandardScaler

```python
print(f"Area: smallest={X_train['area'].min()}, biggest={X_train['area'].max()}")
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("\nAfter scaling:")
print("All features now have similar range!")
print(f"Example - First House's area before scaling: {X_train.values[1][0]}")
print(f"Example - First House's area after scaling: {X_train_scaled[1][0]:.2f}")
```

# train and test model

```python
from sklearn.linear_model import LogisticRegression

# Create model
model = LogisticRegression(max_iter=1000)

# Train the model
model.fit(X_train_scaled, y_train)
print("✅ Model trained!")

y_pred = model.predict(X_test_scaled)
comparison = pd.DataFrame({
    "Actual Price": y_test.values[:5],
    "Predicted Price": y_pred[:5]
})
display(comparison)
```

# Predict a House Price

```python
new_house = pd.DataFrame([{
    'area': 5000,
    'bedrooms': 3,
    'bathrooms': 2,
    'stories': 2,
    'mainroad': 1,
    'guestroom': 0,
    'basement': 1,
    'hotwaterheating': 0,
    'airconditioning': 1,
    'parking': 2,
    'prefarea': 1,
    'furnishingstatus': 0
}])
predicted_price = model.predict(new_house)[0]   # [0] means the result will be shown without []
print("\nPredicted House Price:", predicted_price)
```
