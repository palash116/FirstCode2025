```python
import numpy as np
math_scores = df['math score'].values
reading_scores = df['reading score'].values
writing_scores = df['writing score'].values
```

```python
# 1. Calculate mean, median, std, min, max for math score subject using NumPy

print("Math score- mean:",np.mean(math_scores))
print("Math score- median:",np.median(math_scores))
print("Math score- Std.:",np.std(math_scores))
print("Math score- minimum:",np.min(math_scores))
print("Math score- maximum:",np.max(math_scores))
```

```python
# 1. Calculate mean, median, std, min, max for reading score subject using NumPy

print("Reading score- mean:",np.mean(reading_scores))
print("Reading score- median:",np.median(reading_scores))
print("Reading score- Std.:",np.std(reading_scores))
print("Reading score- minimum:",np.min(reading_scores))
print("Reading score- maximum:",np.max(reading_scores))
```

```python
# 1. Calculate mean, median, std, min, max for writing score subject using NumPy

print("Writing score- mean:",np.mean(writing_scores))
print("Writing score- median:",np.median(writing_scores))
print("Writing score- Std.:",np.std(writing_scores))
print("Writing score- minimum:",np.min(writing_scores))
print("Writing score- maximum:",np.max(writing_scores))
```

```python
# 2. Find total students who scored above 80 in math
above_80=math_scores>80
print("Total students who scored above 80 in math:",np.sum(above_80))
```

```python
# 3. Find total students who scored below 50 in any subject
below_50=(math_scores<50) | (reading_scores<50) | (writing_scores<50)
print("Total students who scored below 50 in any subject:",np.sum(below_50))
```

```python
# 4. Overall average score across all three subjects
average = np.mean([math_scores, reading_scores, writing_scores])
print("Overall average score across all subjects:",average)
```

# **Task-2.2**

```python
# Step 1: Create copy and show original missing values
df_practice = df.copy()
print("Step 1: Original missing values:")
print(df_practice.isnull().sum())
```

```python
# Step 2: Create missing data
import numpy as np
np.random.seed(42)  # For reproducibility (always same result)
```

```python
# Pick 50 random rows for math score to become NaN
missing_math_idx = np.random.choice(df_practice.index, 50, replace=False)
df_practice.loc[missing_math_idx, 'math score'] = np.nan
```

```python
# Pick 30 random rows for writing score to become NaN
missing_writing_idx = np.random.choice(df_practice.index, 30, replace=False)
df_practice.loc[missing_writing_idx, 'writing score'] = np.nan
```

```python
print("\nStep 2: After creating missing values:")
print(df_practice.isnull().sum())
```

```python
##### Step 3: Fill missing math scores with the MEDIAN
median_math = df_practice['math score'].median()
df_practice['math score'].fillna(median_math, inplace=True)
```

```python
# Step 4: Fill missing writing scores with the MEAN
mean_writing = df_practice['writing score'].mean()
df_practice['writing score'].fillna(mean_writing, inplace=True)
```

```python
print("\nStep 3 & 4: After fixing missing values:")
print(df_practice.isnull().sum())
```

```python
# Step 5: Verify
if df_practice.isnull().sum().sum() == 0:
    print("\n All missing values fixed! Dataset is clean again.")
else:
    print("\n Some missing values remain, please check.")
```
