# **Task-1**

```python

print("Unique values in abonoment_type columns:",gym['abonoment_type'].unique())
```

## Bar chart of gym membership types

```python
membership=gym['abonoment_type'].value_counts()
plt.bar(membership.index,membership.values, color=['pink','blue'])
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.title('Gym membership types')
plt.xlabel('abonoment_type')
plt.ylabel("count")
plt.show()
```

## Histogram of time in gym

```python
plt.hist(gym['avg_time_in_gym'])
plt.title('Histogram of time in gym')
plt.show()
```

## Scatter plot: Age vs visits per week

```python
plt.scatter(gym['Age'], gym['visit_per_week'],alpha=0.7)
plt.title('Age vs visit per week')
plt.grid(True)
plt.xlabel('Age')
plt.ylabel('Week')
plt.show()
```

# **Seaborn**

Box plot: Gender vs Age

```python
sns.boxplot(data=gym, x='gender', y='Age')
plt.title('Gender vs Age')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.show()
```

## Count plot: Days per week

```python
sns.countplot(data=gym, x='days_per_week')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.title('Days per week')
plt.show()
```

## Bar plot: Average visits by membership type

```python
sns.barplot(data=gym, x='abonoment_type', y='visit_per_week')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.title('Average visits by membership type')
plt.show()
```
