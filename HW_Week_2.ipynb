# **Home work-2.12**

```python
####  Write a program that prints a simple number pyramid pattern using nested loops.

n=5
for i in range(1,n+1):
    for j in range(n-i):
        print(" ", end="")
    for k in range(2*i-1):
        print("*", end="")
    print()
```
